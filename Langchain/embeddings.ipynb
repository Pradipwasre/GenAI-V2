```python
from langchain_community.document_loaders import WebBaseLoader
from langchain_text_splitters import RecursiveCharacterTextSplitter
from langchain_huggingface import HuggingFaceEmbeddings
from langchain_core.vectorstores import InMemoryVectorStore

# 1. Load cricket website via HTML parser
loader = WebBaseLoader(
    web_path="https://en.wikipedia.org/wiki/Cricket",
    default_parser="html.parser",
)
docs = loader.load()

# 2. Split into chunks using HTML-aware separators
splitter = RecursiveCharacterTextSplitter.from_language(
    language="html",
    chunk_size=1000,
    chunk_overlap=200,
)
chunks = splitter.split_documents(docs)

# 3. Initialize HuggingFace embedding model (local, no API key)
embeddings = HuggingFaceEmbeddings(
    model_name="sentence-transformers/all-MiniLM-L6-v2",
    model_kwargs={"device": "cpu"},
    encode_kwargs={"normalize_embeddings": False},
)

# 4. Store chunks + embeddings in RAM
vector_store = InMemoryVectorStore(embedding=embeddings)
vector_store.add_documents(documents=chunks)

# 5. Verify: inspect stored vectors
print("Total chunks stored:", len(chunks))
print("Sample chunk text:", chunks[0].page_content[:200])

# Peek into the in-memory store
for idx, (doc_id, entry) in enumerate(vector_store.store.items()):
    if idx >= 3:
        break
    print("Doc ID:", doc_id)
    print("Vector dim:", len(entry["vector"]))
    print("Text preview:", entry["text"][:150])
    print("---")
```

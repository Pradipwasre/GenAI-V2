# Regular Expressions in Python Notes

This notebook provides notes and examples on using regular expressions (regex) in Python.

## What is Regular Expression?

A regular expression is a sequence of characters that defines a search pattern. When you search for a particular pattern in a text, you can use regular expressions to describe what you are looking for. They are powerful tools for string manipulation, pattern matching, and data validation.

## How Regular Expressions Work

Regular expressions work by matching a sequence of characters against a defined pattern. The Python `re` module provides functions to perform these operations. When a regex engine processes a pattern, it attempts to find an occurrence of that pattern within a target string. If a match is found, the engine returns information about the match; otherwise, it indicates that no match was found.

## Types of Regular Expression Patterns

Regular expressions use various special characters and sequences to define patterns:

*   **Literal Characters**: Match themselves (e.g., `a`, `1`, `!`).
*   **Metacharacters**: Special characters with predefined meanings.
    *   `.`: Matches any character (except newline).
    *   `^`: Matches the start of the string.
    *   `$`: Matches the end of the string.
    *   `*`: Matches zero or more occurrences of the preceding character.
    *   `+`: Matches one or more occurrences of the preceding character.
    *   `?`: Matches zero or one occurrence of the preceding character.
    *   `{n}`: Matches exactly `n` occurrences.
    *   `{n,m}`: Matches between `n` and `m` occurrences.
    *   `|`: Acts as an OR operator.
    *   `()`: Groups expressions and creates capturing groups.
    *   `[]`: Defines a character set (e.g., `[abc]` matches 'a', 'b', or 'c').
    *   `\`: Escapes special characters or denotes special sequences.
*   **Special Sequences**: Predefined character classes.
    *   `\d`: Matches any digit (0-9).
    *   `\D`: Matches any non-digit character.
    *   `\w`: Matches any word character (alphanumeric and underscore).
    *   `\W`: Matches any non-word character.
    *   `\s`: Matches any whitespace character.
    *   `\S`: Matches any non-whitespace character.

## Rules for Regular Expressions

1.  **Specificity**: Be as specific as possible to avoid unintended matches.
2.  **Escaping**: Use a backslash (`\`) to escape metacharacters if you want to match them literally (e.g., `\.` to match a literal dot).
3.  **Raw Strings**: In Python, use raw strings (prefix with `r`, e.g., `r'\d+'`) for regular expressions to avoid issues with backslash escapes being interpreted by Python itself.
4.  **Greediness**: Quantifiers (`*`, `+`, `?`, `{}`) are by default 'greedy', meaning they will match the longest possible string. Add `?` after them to make them 'non-greedy' (e.g., `*?`).
5.  **Anchors**: Use `^` and `$` to anchor patterns to the beginning or end of a string (or lines in multiline mode) for precise matches.

## Demonstration: Regex on a Paragraph

Here is a sample paragraph to demonstrate various regex operations. This paragraph contains several words, numbers, and punctuation marks. You can search for specific patterns like email addresses, phone numbers (though none are explicitly here, we can simulate), or just common words. The goal is to illustrate how flexible and powerful regular expressions are for extracting information from text data. Remember, Python's `re` module is your friend for these tasks.

```python
import re

paragraph = "Here is a sample paragraph to demonstrate various regex operations. This paragraph contains several words, numbers, and punctuation marks. You can search for specific patterns like email addresses, phone numbers (though none are explicitly here, we can simulate), or just common words. The goal is to illustrate how flexible and powerful regular expressions are for extracting information from text data. Remember, Python's re module is your friend for these tasks."
```

```python
# Find all words in the paragraph
words = re.findall(r'\b\w+\b', paragraph)
```

### Regex Pattern Explanation

`words = re.findall(r'\b\w+\b', paragraph)`

- **`\b`** → Word boundary, ensures we capture whole words rather than fragments.
- **`\w+`** → Matches one or more word characters:
  - Letters (A–Z, a–z)
  - Digits (0–9)
  - Underscore (`_`)
- **`\b`** → Word boundary at the end, so the match stops cleanly.

```python
words
```

```python
print("All words:", words[:10], "...") # Display first 10 words
```

```python
# Find all occurrences of 'paragraph'
paragraph_matches = re.findall(r'paragraph', paragraph)
print("Occurrences of 'paragraph':", len(paragraph_matches))
```

```python
# Find words that start with 's' or 'S'
paragraph = "Sam sells snacks in the small shop."
```

```python
starts_with_s = re.findall(r'\b[sS]\w*\b', paragraph)
```

### Regex Pattern Explanation

`starts_with_s = re.findall(r'\b[sS]\w*\b', paragraph)`

- **`\b`** → Word boundary, ensures we match whole words.
- **`[sS]`** → Matches either lowercase `s` or uppercase `S` at the start.
- **`\w*`** → Matches zero or more word characters (letters, digits, underscore) following the `s`/`S`.
- **`\b`** → Word boundary at the end, so the match stops at the word’s end.

```python
starts_with_s
```

## Different Regex Examples

```python
import re

text = "Contact us at info@example.com or support@company.org."
```

```python
pattern = r'\b[A-Za-z0-9._%+-]+@[A-Za-z0-9.-]+\.[A-Za-z]{2,}\b'
```

### Regex Pattern Explanation

`pattern = r'\b[A-Za-z0-9._%+-]+@[A-Za-z0-9.-]+\.[A-Za-z]{2,}\b'`

- **`\b`** → Word boundary, ensures we match whole words (not partial text).
- **`[A-Za-z0-9._%+-]+`** → Matches the username part of the email:
  - Letters (A–Z, a–z)
  - Digits (0–9)
  - Special characters: `. _ % + -`
  - `+` means one or more of these characters.
- **`@`** → Literal `@` symbol separating username and domain.
- **`[A-Za-z0-9.-]+`** → Matches the domain name:
  - Letters, digits, dot (`.`), and hyphen (`-`).
- **`\.`** → Literal dot before the top-level domain.
- **`[A-Za-z]{2,}`** → Matches the top-level domain (like `com`, `org`, `info`):
  - Only letters allowed.
  - `{2,}` means at least 2 characters.
- **`\b`** → Word boundary at the end.

```python
emails = re.findall(pattern, text)
```

```python
print("Emails found:", emails)
```

```python
# Example 2: Validating a simple date format (YYYY-MM-DD)

import re

date_string1 = "Today's date is 2023-10-26."
date_string2 = "Invalid date: 2023/10/26."
```

```python
date_pattern = r'\d{4}-\d{2}-\d{2}'
```

### Regex Pattern Explanation

`date_pattern = r'\d{4}-\d{2}-\d{2}'`

- **`\d{4}`** → Matches exactly 4 digits (the year).
- **`-`** → Literal dash separating parts of the date.
- **`\d{2}`** → Matches exactly 2 digits (month).
- **`-`** → Another literal dash.
- **`\d{2}`** → Matches exactly 2 digits (day).

```python
print("Date in string 1:", re.findall(date_pattern, date_string1))
```

```python
print("Date in string 2:", re.findall(date_pattern, date_string2))
```

```python
import re

# Example 3: Extracting numbers from a string
measurements = "The temperature was 25.5 degrees Celsius, and the pressure was 1012.3 hPa."
```

```python
numbers = re.findall(r'\d+\.?\d*', measurements)
```

```python
numbers
```

### Regex Pattern Explanation

`numbers = re.findall(r'\d+\.?\d*', measurements)`

- **`\d+`** → Matches one or more digits (the integer part).
- **`\.?`** → Matches an optional decimal point (`.`).
- **`\d*`** → Matches zero or more digits after the decimal (the fractional part).
- Together, this captures both whole numbers and decimal numbers.
